```python
from gtts import gTTS
import IPython.display as ipd
text = "Welcome to Day 20 of AI Training. Today we learn Speech AI!"
tts = gTTS(text=text, lang='en')
tts.save("harika.mp3")
ipd.Audio("harika.mp3")
```

**Speech-to-Text(STT)**

```python
!pip install git+https://github.com/openai/whisper.git -q
!apt-get install ffmpeg -y -q
import whisper
```

```python
model = whisper.load_model("base")
```

```python
result = model.transcribe("harika.mp3", fp16=False)
```

```python
print("--- TRANSCRIPTION RESULTS ---")
print("Detected Language:", result.get("language", "en").upper())
print("Full Text:", result["text"].strip())
print("\n--- TIMED SEGMENTS ---")
for seg in result["segments"]:
  print(f"[{seg['start']:.1f}s -> {seg['end']:.1f}s]: {seg['text'].strip()}")
```
